```python
from pyfaidx import Fasta
import pandas as pd
import re
```

```python
Acer_gtf = pd.read_csv("./gtf/Apis_cerana.tsv", sep="\t")
Acer_gtf
```

```python
Acer_pid_gid = Acer_gtf[["protein_id", "gene_id"]]
Acer_pid_gid
```

```python
Acer_pid_gid[Acer_pid_gid["protein_id"].isna() | Acer_pid_gid["gene_id"].isna()]
```

```python
Acer_faa = Fasta("./all_former_faa_origin/GCF_029169275.1_AcerK_1.0_protein.faa", read_long_names=True)
header_set = set(long_name.split(' ')[0] for long_name in Acer_faa.keys())
header_set
```

faa 中的 header 都会出现在 GTF 的 protein_id 列中吗？

```python
pid_set = set(Acer_pid_gid["protein_id"])
print(len(pid_set))
print(len(header_set))
print(len(pid_set & header_set))
print(pid_set - header_set)
```

是的。faa 中的 header 都会出现在 GTF 的 protein_id 列中

```python
with open("./corrected/Acer_corrected.faa", "w") as fout:
    for long_name in Acer_faa.keys():
        pid_short_header = long_name.split(' ')[0]
        gid = Acer_pid_gid.loc[Acer_pid_gid["protein_id"] == pid_short_header, "gene_id"].iloc[0]
        new_header = long_name.replace(pid_short_header, gid, 1)
        seq = Acer_faa[long_name][:].seq

        fout.write(f">{new_header}\n")
        fout.write(seq + "\n")
```

```python
from pyfaidx import Fasta

Aaeg_faa = Fasta("./orthofinder_output/Aedes_aegypti.faa")
Aaeg_faa_header_list = list(Aaeg_faa.keys())
Aaeg_faa_header_list
```

结果显示 primary_transcript 处理后的 faa 中无重复 header
